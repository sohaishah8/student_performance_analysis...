```python
import numpy as np
students=np.array([
      'iqbal',
       'ali',
      'ahmed',
     'memood',
     'sadeem'
     ])
subject=np.array([
      'math',
     'physics',
     'python',
    'english',
    'chemistry'
        ])
marks=np.array([
   [87,10,45,78],
   [45,4,51,98],
   [43,91,76,61],
   [41,87,54,71],
   [51,63,87,71]
   ])
print("/////======students performance analiyes system======/////")
print('student name:')
print(students)

print('--\nsubject')
print(subject)

print('----\noriginal marks')
print(marks)

bounce=np.array([5,6,9,8])
final_marks=marks+bounce

total_marks=np.sum(final_marks,axis=1)

print('\ntotal marks')
for i in range(len(students)):
        print(f'{students[i]} : {total_marks[i]}')

average_marks=np.mean(final_marks,axis=1)
print('\naverage marks')
for i in range(len(students)):
     print(f'{students[i]} : round({average_marks[i]},2)')

highest_index=np.argmax(total_marks)
print('----\nhighest performance:')
print(students[highest_index])
print('total marks:',total_marks[highest_index])

lowest_index=np.argmin(total_marks)
print('---\nlowest performance:')
print(students[lowest_index])
print('total marks: ',total_marks[lowest_index])

passing_marks=33
print('\n========RESULT=====')
for i in range(len(students)):

     if average_marks[i] >= passing_marks:
         result='CONGRATULATION YOU ARE PASS'
     else:
         result='try again(fail)'
     print(f'{students[i]}||TOTAL:{total_marks[i]}||Average marks:{round(average_marks[i],2)}||{result}')
      
```
